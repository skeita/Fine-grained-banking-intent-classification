## Frozen DistilBERT — E2

**What this does:** keeps the pretrained encoder frozen and trains only the classification head.

**Why:** this isolates the value of the pretrained representation without allowing the encoder to adapt to banking language.


```python
# E2 is a transfer-learning control: only the new 77-class head is trainable.
# The validation set selects the best epoch; the official test set is evaluated later.
# E2 measures the value of pretrained representations when the encoder stays frozen.
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))
import json
from src.data.load_data import load_banking77
from src.models.transformer import run_transformer_experiment
dataset = load_banking77(validation_size=0.1, seed=42)
result = run_transformer_experiment(dataset, frozen=True, learning_rate=2e-05, epochs=3, max_length=64, seed=42)
print({"history": result["history"], "training_seconds": result["training_seconds"], "trainable_parameters": result["trainable_parameters"], "total_parameters": result["total_parameters"]})
Path("results/logs").mkdir(parents=True, exist_ok=True)
Path("results/logs/04_distilbert_frozen_history.json").write_text(json.dumps({"history": result["history"], "training_seconds": result["training_seconds"], "trainable_parameters": result["trainable_parameters"], "total_parameters": result["total_parameters"]}, indent=2))
result["model"].save_pretrained("results/04_distilbert_frozen_checkpoint")
result["tokenizer"].save_pretrained("results/04_distilbert_frozen_checkpoint")
```

## Training curves — E2 Frozen DistilBERT

**What this does:** displays the saved training/validation loss and validation Macro-F1.

**Why:** curves show whether learning is progressing, whether the model is overfitting, and how the best checkpoint was selected.

```python
# Read the saved history produced by the training step.
from pathlib import Path
import json
import pandas as pd
import matplotlib.pyplot as plt

root = Path.cwd().parent
history_file = root / "results/logs/04_distilbert_frozen_history.json"
history_data = json.loads(history_file.read_text())
history = history_data if isinstance(history_data, list) else history_data["history"]
frame = pd.DataFrame(history)
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
axes[0].plot(frame["epoch"], frame["train_loss"], marker="o", label="train loss")
axes[0].plot(frame["epoch"], frame["validation_loss"], marker="o", label="validation loss")
axes[0].set_title("E2 Frozen DistilBERT loss")
axes[0].set_xlabel("Epoch")
axes[0].legend()
axes[1].plot(frame["epoch"], frame["validation_macro_f1"], marker="o", color="#55A868")
axes[1].set_title("E2 Frozen DistilBERT validation Macro-F1")
axes[1].set_xlabel("Epoch")
axes[1].set_ylim(0, 1)
plt.tight_layout()
plt.show()
```
